# STEP 1 — Build Recipe Dataset
Menggabungkan semua CSV resep per-protein menjadi satu dataframe bersih.

**Input:**
- `data/resep/dataset-ayam.csv`
- `data/resep/dataset-ikan.csv`
- `data/resep/dataset-kambing.csv`
- `data/resep/dataset-sapi.csv`
- `data/resep/dataset-tahu.csv`
- `data/resep/dataset-telur.csv`
- `data/resep/dataset-tempe.csv`
- `data/resep/dataset-udang.csv`

**Output:** `data/resep_clean.csv`

```python
import pandas as pd
import glob
import os
import re
```

```python
# ============================================================
# LOAD SEMUA CSV RESEP
# ============================================================

datafolder_path = "../data/resep/*.csv"
filescsv = glob.glob(datafolder_path)
print(f"File ditemukan: {len(filescsv)}")
for f in filescsv:
    print(" -", os.path.basename(f))
```

```python
# ============================================================
# GABUNGKAN + TAMBAH KOLOM KATEGORI
# ============================================================

all_df = []

for file in filescsv:
    df = pd.read_csv(file)
    judul_file = os.path.basename(file)                     # contoh: dataset-ayam.csv
    kategori = judul_file.replace("dataset-", "")           # contoh: ayam.csv
    df["Kategori"] = kategori
    all_df.append(df)

df_resep = pd.concat(all_df, ignore_index=True)
print(f"Total resep tergabung: {len(df_resep)}")
print(df_resep["Kategori"].value_counts())
```

```python
# ============================================================
# CLEANING FUNCTION
# ============================================================

def clean_text(text):
    """Lowercase, hapus karakter khusus, rapikan spasi."""
    text = str(text).lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
# ============================================================
# FILTER DATA TIDAK VALID
# ============================================================

# Hapus baris yang Ingredients atau Steps kosong
df_resep = df_resep.dropna(subset=["Ingredients", "Steps"])

# Bersihkan judul
df_resep["clean_title"] = df_resep["Title"].apply(clean_text)

# Hapus judul kosong / placeholder
invalid_titles = {"", "notitle", "[notitle]"}
df_resep = df_resep[~df_resep["clean_title"].isin(invalid_titles)]
df_resep = df_resep[df_resep["clean_title"].str.len() >= 3]

print(f"Resep valid setelah filter: {len(df_resep)}")
```

```python
# ============================================================
# DEDUPLIKASI — PERTAHANKAN STEPS TERPANJANG
# ============================================================

df_resep["steps_length"] = df_resep["Steps"].fillna("").astype(str).apply(len)

df_resep = df_resep.sort_values("steps_length", ascending=False)
df_resep = df_resep.drop_duplicates(subset="clean_title", keep="first")
df_resep = df_resep.reset_index(drop=True)

print(f"Resep unik setelah deduplikasi: {len(df_resep)}")
```

```python
# ============================================================
# SIMPAN
# ============================================================

output_cols = ["Title", "Ingredients", "Steps", "Loves", "URL", "Kategori", "clean_title", "steps_length"]
df_resep[output_cols].to_csv("../data/resep_clean.csv", index=False)

print(f"\nSaved: data/resep_clean.csv")
print(f"Total: {len(df_resep)} resep")
df_resep[["Title", "Kategori", "steps_length"]].head(5)
```
