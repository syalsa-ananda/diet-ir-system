# STEP 4 — Evaluasi Sistem IR (Revisi)

Perbaikan dari versi sebelumnya:
1. **Query set lebih variatif** — bahasa natural user, bukan hanya nama kategori
2. **Fungsi relevansi lebih adil** — kalori masuk sebagai kriteria utama relevansi
3. **Popularity boost** — tambahkan komponen `Loves` ke hybrid score
4. **Weight tuning** — cari bobot terbaik secara sistematis
5. **Perbandingan 6 kondisi** — dari baseline hingga hybrid terbaik

**Input:** `src/ir_model.pkl`

**Output:** `output/evaluation_results.csv`, `output/weight_tuning_results.csv`

```python
import sys
sys.path.append("../src")

import pandas as pd
import numpy as np
import pickle
import os
import re

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import ndcg_score
from rank_bm25 import BM25Okapi
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

print("Library siap.")
```

```python
# ============================================================
# LOAD MODEL
# ============================================================

with open("../src/ir_model.pkl", "rb") as f:
    bundle = pickle.load(f)

tfidf_vectorizer = bundle["tfidf_vectorizer"]
tfidf_matrix     = bundle["tfidf_matrix"]
bm25_index       = bundle["bm25_index"]
df               = bundle["df"].reset_index(drop=True)
KALORI_THRESHOLD = bundle.get("calorie_threshold", 500.0)

print(f"Model loaded. Corpus: {len(df):,} dokumen")
```

```python
# ============================================================
# PREPROCESSING
# ============================================================

stemmer          = StemmerFactory().create_stemmer()
stopword_remover = StopWordRemoverFactory().create_stop_word_remover()

STOPWORDS_TAMBAHAN = {
    "resep","cara","membuat","masak","memasak","buat",
    "mudah","enak","lezat","sedap","nikmat",
    "praktis","simpel","simple","cepat","yummy","mantap",
}

KATA_DIET = [
    "diet","rendah kalori","rendah lemak","sehat","ringan",
    "tidak gemuk","nggak gemuk","gak gemuk","tidak bikin gemuk",
    "langsing","kurus","turun berat","berat badan",
    "low calorie","low fat","light","kalori rendah",
    "ga bikin gemuk","ga bikin buncit","buncit",
]


def preprocess_text(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    text = stopword_remover.remove(text)
    text = stemmer.stem(text)
    return " ".join(w for w in text.split() if w not in STOPWORDS_TAMBAHAN)


def deteksi_diet(query):
    return any(k in query.lower() for k in KATA_DIET)


# Hitung skor komponen (sekali saja)
def cal_score(cal):
    if pd.isna(cal) or cal <= 0:
        return 0.5
    return round(1.0 - min(cal / KALORI_THRESHOLD, 1.0), 4)


calorie_scores    = df["calories"].apply(cal_score).values
loves             = pd.to_numeric(df["Loves"], errors="coerce").fillna(0).values.astype(float)
loves_max         = loves.max()
popularity_scores = loves / loves_max if loves_max > 0 else loves

print("Preprocessing & skor komponen siap.")
```

```python
# ============================================================
# FUNGSI SEARCH
# ============================================================

def search(query, top_k, alpha, beta, gamma, delta=0.0, adaptive=False, kalori_max=None):
    q = preprocess_text(query)
    if not q.strip():
        return pd.DataFrame()

    # TF-IDF
    ts = cosine_similarity(tfidf_vectorizer.transform([q]), tfidf_matrix)[0]
    tn = ts / ts.max() if ts.max() > 0 else ts

    # BM25
    bs = bm25_index.get_scores(q.split())
    bn = bs / bs.max() if bs.max() > 0 else bs

    # Adaptive weights
    if adaptive and deteksi_diet(query):
        a = alpha * 0.875
        b = beta  * 0.875
        g = min(gamma + 0.05, 0.35)
        d = delta
    else:
        a, b, g, d = alpha, beta, gamma, delta

    # Normalisasi bobot
    tw = a + b + g + d
    if tw > 0:
        a, b, g, d = a/tw, b/tw, g/tw, d/tw

    hybrid = a * tn + b * bn + g * calorie_scores + d * popularity_scores

    if kalori_max is not None:
        mask   = (df["calories"].notna() & (df["calories"] <= kalori_max)).values
        hybrid = hybrid * mask

    top_idx = np.argsort(hybrid)[::-1][:top_k]
    hasil   = df.iloc[top_idx].copy()
    hasil["hybrid_score"] = hybrid[top_idx]
    return hasil[hasil["hybrid_score"] > 0].reset_index(drop=True)


print("Fungsi search siap.")
```

```python
# ============================================================
# QUERY SET — VARIATIF & REALISTIS
# Menggunakan lambda untuk fungsi relevansi per query
# ============================================================

BATAS_RENDAH = 300
BATAS_SEDANG = 400
BATAS_NORMAL = 500

QUERY_SET = [
    {
        "query":   "olahan ayam yang tidak bikin gemuk",
        "desc":    "ayam rendah kalori",
        "relevan": lambda r: "ayam" in str(r.get("Kategori","")).lower() and pd.notna(r.get("calories")) and r["calories"] <= BATAS_SEDANG,
    },
    {
        "query":   "ikan kukus atau bakar sehat buat diet",
        "desc":    "ikan rendah kalori",
        "relevan": lambda r: "ikan" in str(r.get("Kategori","")).lower() and pd.notna(r.get("calories")) and r["calories"] <= BATAS_SEDANG,
    },
    {
        "query":   "tahu tempe yang enak tapi rendah kalori",
        "desc":    "tahu tempe rendah kalori",
        "relevan": lambda r: any(k in str(r.get("Kategori","")).lower() for k in ["tahu","tempe"]) and pd.notna(r.get("calories")) and r["calories"] <= BATAS_RENDAH,
    },
    {
        "query":   "masakan udang rendah lemak untuk makan malam",
        "desc":    "udang rendah kalori",
        "relevan": lambda r: "udang" in str(r.get("Kategori","")).lower() and pd.notna(r.get("calories")) and r["calories"] <= BATAS_SEDANG,
    },
    {
        "query":   "sarapan sehat pakai telur kalori sedikit",
        "desc":    "telur rendah kalori",
        "relevan": lambda r: "telur" in str(r.get("Kategori","")).lower() and pd.notna(r.get("calories")) and r["calories"] <= BATAS_RENDAH,
    },
    {
        "query":   "sapi rebus atau panggang yang tidak berlemak",
        "desc":    "sapi rendah kalori",
        "relevan": lambda r: "sapi" in str(r.get("Kategori","")).lower() and pd.notna(r.get("calories")) and r["calories"] <= BATAS_NORMAL,
    },
    {
        "query":   "lauk pauk diet rendah kalori buat makan siang",
        "desc":    "semua kategori rendah kalori",
        "relevan": lambda r: pd.notna(r.get("calories")) and r["calories"] <= BATAS_SEDANG,
    },
    {
        "query":   "makanan berprotein tinggi kalori rendah",
        "desc":    "protein tinggi kalori rendah",
        "relevan": lambda r: pd.notna(r.get("calories")) and r["calories"] <= BATAS_SEDANG and pd.notna(r.get("proteins")) and r["proteins"] >= 15.0,
    },
    {
        "query":   "masakan ayam buat yang lagi diet nggak mau gemuk",
        "desc":    "ayam diet informal",
        "relevan": lambda r: "ayam" in str(r.get("Kategori","")).lower() and pd.notna(r.get("calories")) and r["calories"] <= BATAS_SEDANG,
    },
    {
        "query":   "ikan yang ga bikin buncit cocok buat diet",
        "desc":    "ikan diet informal",
        "relevan": lambda r: "ikan" in str(r.get("Kategori","")).lower() and pd.notna(r.get("calories")) and r["calories"] <= BATAS_SEDANG,
    },
]

TOP_K = 10

print(f"Query set: {len(QUERY_SET)} query | Top-K: {TOP_K}")
print()
print("Distribusi ground truth per query:")
print(f"  {'No':<4} {'Desc':<35} {'Relevan':>10}")
print("  " + "-" * 52)
for i, q in enumerate(QUERY_SET, 1):
    total = sum(1 for _, r in df.iterrows() if q["relevan"](r))
    flag  = " ⚠ sedikit" if total < 5 else ""
    print(f"  Q{i:02d}  {q['desc']:<35} {total:>6} dok{flag}")
```

```python
# ============================================================
# FUNGSI METRIK
# ============================================================

def precision_at_k(rel, k):        return sum(rel[:k]) / k
def recall_at_k(rel, tot, k):      return sum(rel[:k]) / min(tot, k) if tot > 0 else 0.0
def mrr_score(rel):                return next((1/(i+1) for i,r in enumerate(rel) if r), 0.0)
def ndcg_at_k(rel, k):
    r = np.array(rel[:k], dtype=float).reshape(1,-1)
    i = np.sort(r, axis=1)[:,::-1]
    return ndcg_score(i, r) if i.sum() > 0 else 0.0

print("Fungsi metrik siap.")
```

```python
# ============================================================
# FUNGSI EVALUASI
# ============================================================

def evaluasi(nama, alpha, beta, gamma, delta=0.0, adaptive=False, verbose=True):
    rows = []
    for q in QUERY_SET:
        total_rel = min(sum(1 for _, r in df.iterrows() if q["relevan"](r)), 50)
        hasil = search(q["query"], TOP_K, alpha, beta, gamma, delta, adaptive)

        if hasil is None or len(hasil) == 0:
            rows.append({"query": q["desc"], "P@K": 0, "R@K": 0, "MRR": 0, "NDCG": 0})
            continue

        rel = [1 if q["relevan"](r) else 0 for _, r in hasil.iterrows()]
        rel += [0] * (TOP_K - len(rel))

        rows.append({
            "query": q["desc"],
            "P@K":   round(precision_at_k(rel, TOP_K), 4),
            "R@K":   round(recall_at_k(rel, total_rel, TOP_K), 4),
            "MRR":   round(mrr_score(rel), 4),
            "NDCG":  round(ndcg_at_k(rel, TOP_K), 4),
        })

    df_e = pd.DataFrame(rows)
    if verbose:
        print(f"\n{'='*65}")
        print(f"  {nama}  (α={alpha:.2f} β={beta:.2f} γ={gamma:.2f} δ={delta:.2f})")
        print(f"{'='*65}")
        print(df_e.to_string(index=False))
        print(f"  Precision@{TOP_K} : {df_e['P@K'].mean():.4f}")
        print(f"  Recall@{TOP_K}    : {df_e['R@K'].mean():.4f}")
        print(f"  MRR          : {df_e['MRR'].mean():.4f}")
        print(f"  NDCG@{TOP_K}     : {df_e['NDCG'].mean():.4f}")
    return df_e


print("Fungsi evaluasi siap.")
```

```python
# ============================================================
# EVALUASI 5 KONDISI
# ============================================================

df_tfidf         = evaluasi("TF-IDF Saja",                  alpha=1.0, beta=0.0, gamma=0.0)
df_bm25          = evaluasi("BM25 Saja",                    alpha=0.0, beta=1.0, gamma=0.0)
df_no_cal        = evaluasi("TF-IDF + BM25 (tanpa kalori)", alpha=0.5, beta=0.5, gamma=0.0)
df_hybrid        = evaluasi("Hybrid + Calorie",             alpha=0.4, beta=0.4, gamma=0.2, adaptive=True)
df_hybrid_pop    = evaluasi("Hybrid + Calorie + Popularity",alpha=0.35,beta=0.35,gamma=0.2, delta=0.1, adaptive=True)
```

```python
# ============================================================
# WEIGHT TUNING
# ============================================================

print("Menjalankan weight tuning...\n")

kandidat = [
    (0.5, 0.3, 0.2, 0.0),
    (0.4, 0.4, 0.2, 0.0),
    (0.3, 0.5, 0.2, 0.0),
    (0.4, 0.3, 0.3, 0.0),
    (0.3, 0.4, 0.3, 0.0),
    (0.3, 0.3, 0.4, 0.0),
    (0.35, 0.35, 0.2, 0.1),
    (0.30, 0.30, 0.3, 0.1),
    (0.25, 0.35, 0.3, 0.1),
    (0.30, 0.30, 0.2, 0.2),
]

tuning = []
for a, b, g, d in kandidat:
    ev = evaluasi(f"a={a} b={b} g={g} d={d}", alpha=a, beta=b, gamma=g, delta=d, adaptive=True, verbose=False)
    tuning.append({"alpha": a, "beta": b, "gamma": g, "delta": d,
                   "P@K": ev["P@K"].mean(), "R@K": ev["R@K"].mean(),
                   "MRR": ev["MRR"].mean(), "NDCG": ev["NDCG"].mean()})

df_tuning = pd.DataFrame(tuning).sort_values("NDCG", ascending=False).round(4)

print("Hasil Weight Tuning (diurutkan NDCG):")
print(df_tuning.to_string(index=False))

best = df_tuning.iloc[0]
print(f"\nBobot terbaik: α={best['alpha']}, β={best['beta']}, γ={best['gamma']}, δ={best['delta']}")
print(f"NDCG          : {best['NDCG']:.4f}")
```

```python
# ============================================================
# EVALUASI DENGAN BOBOT TERBAIK
# ============================================================

df_best = evaluasi(
    f"Hybrid BEST (α={best['alpha']} β={best['beta']} γ={best['gamma']} δ={best['delta']})",
    alpha=float(best["alpha"]), beta=float(best["beta"]),
    gamma=float(best["gamma"]), delta=float(best["delta"]),
    adaptive=True
)
```

```python
# ============================================================
# TABEL PERBANDINGAN FINAL
# ============================================================

def r(df_e, m): return round(df_e[m].mean(), 4)

tabel = pd.DataFrame({
    "Metrik":              [f"Precision@{TOP_K}", f"Recall@{TOP_K}", "MRR", f"NDCG@{TOP_K}"],
    "TF-IDF":              [r(df_tfidf, m)      for m in ["P@K","R@K","MRR","NDCG"]],
    "BM25":                [r(df_bm25, m)        for m in ["P@K","R@K","MRR","NDCG"]],
    "TF-IDF+BM25":         [r(df_no_cal, m)      for m in ["P@K","R@K","MRR","NDCG"]],
    "Hybrid+Calorie":      [r(df_hybrid, m)      for m in ["P@K","R@K","MRR","NDCG"]],
    "Hybrid+Cal+Pop":      [r(df_hybrid_pop, m)  for m in ["P@K","R@K","MRR","NDCG"]],
    "Hybrid BEST":         [r(df_best, m)        for m in ["P@K","R@K","MRR","NDCG"]],
})

print("\n" + "="*90)
print("  TABEL PERBANDINGAN METRIK EVALUASI (REVISI)")
print("="*90)
print(tabel.to_string(index=False))

print("\nKolom terbaik per metrik:")
for _, row in tabel.iterrows():
    nilai    = row.drop("Metrik")
    best_col = nilai.idxmax()
    print(f"  {row['Metrik']:<15} → {best_col} ({nilai[best_col]:.4f})")

os.makedirs("../output", exist_ok=True)
tabel.to_csv("../output/evaluation_results.csv", index=False)
df_tuning.to_csv("../output/weight_tuning_results.csv", index=False)

print("\nSaved:")
print("  output/evaluation_results.csv")
print("  output/weight_tuning_results.csv")
```
